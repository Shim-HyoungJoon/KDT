```python
# 커피 자판기 음료수 품절 시

coffee_stock = 10

while coffee_stock > 0 :
    print(f'커피가 준비되었습니다.')
    coffee_stock -= 1
    print(f'남은 재고: {coffee_stock}잔')
   
print('커피 재고가 모두 소진되었습니다. 판매를 종료합니다.')
```

```python
customer_stamps = [5, 12, 7, 15, 3]      # 손님들이 모든 스탬프 개수

for stamp in customer_stamps:
    if stamp >= 10:
        print(f'스탬프 {stamp}개 보유 중! VIP 쿠폰 발급 대상입니다.')
    else:
        print(f'스탬프 {stamp}개 보유 중! 조금만 더 모아보세요!')
```
